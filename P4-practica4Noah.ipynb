##### 2.- Crea dos archivos. Uno almacena los accession que empiezen por 'a' y el otro el resto

```python
accs = ['ab56', 'bh84', 'hv76', 'ay93', 'ap97', 'bd72']
```

```python
archivo_a = open("archivo_a","w+")
archivo_b = open("archivo_b","w+")

for i in accs:
    if i[0] == "a":
        archivo_a.write(i + " ")
    if i[0] != "a":
        archivo_b.write(i + " ")
    archivo_a.seek(0)
    archivo_b.seek(0)
    print("los que empiezan por a:", archivo_a.readlines(), "los que empiezan por otra:", archivo_b.readlines())
```

##### ¿Puedes hacer que devuelva dos decimales? ¿Y que lidie con mayúsculas y minúsculas?

```python
seq = "ATGCAAGTCAAGTAA"
def porcentaje_a(seq):
    seq = seq.upper()
    a_count = seq.count("A")
    porcentaje = (a_count / len(seq)) * 100
    return round(porcentaje, 2)
print(porcentaje_a(seq))
```
