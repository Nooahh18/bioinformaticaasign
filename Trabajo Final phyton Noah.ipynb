#### 5.- Devuelve la secuencia de ARN que surgiría a partir de la siguiente secuencia. Recuerda que la secuencia de ARN resultante tiene que ser complementaria a la de ADN. (1 pto)

```python
def ARNseq(secuencia):
    secuencia="".join(secuencia).upper()  #recorremos cada base de la secuencia y la transformamos para que se lea de una y no haya fallos
    ARN = "" #creamos una lista vacia para hacer el ARN
    for base in secuencia: #aquí hacemos el cambio de bases correspondiente al ARN
        if base == "A": 
            ARN += "T" #A por T
        elif base == "T":
            ARN += "U" #T pot U
        elif base == "G":
            ARN += "C" #G por C
        elif base == "C":
            ARN += "G" #C por G
    return ARN #que devuelva el ARN cambiado
```

```python
arn_seq = ARNseq(secuencia) #lo guardamis en una variable
```

#### 6.- Realiza una función que realice una digestión con las siguientes enzimas de restricción en nuestra secuencia. La función tiene que devolver la siguiente información: (2 pto)
1. Enzimas que tienen sitios de corte en la secuencia y posición (o posiciones de corte)
2. Enzimas que no tienen sitios de corte (si las hay)
3. Resultado de la digestión: los fragmentos resultantes de nuestra secuencia
4. Indicar número y tamaño de cada fragmento

```python
enzR = {
    'HindIII' : 'A*AGCTT',
    'BamHI' : 'G*GATCC',
    'AluI' : 'AG*CT',
    'Sau3AI' : '*GATC',
    'EcoRI' : 'G*AATTC'
}
```

```python
def digestion_restriccion(secuencia, enzR):
    secuencia_str = ''.join(secuencia).upper() #convertimos la secuencia en una cadega continua
    
    enzimas_cortan = {} #creamos un diccionario vacio donde guardamos las encimas que cortan y la posición en que lo hacen
    enzimas_no_cortan = [] #creamos una lista para guardar las que no cortan

    for enzima, sitio in enzR.items():  # recorremos cada enzima y su sitio de corte
        secuencia_recon = sitio.replace('*', '') # quitamos el asterisco que indica el punto de corte
        posiciones = []  # lista para guardar las posiciones donde corta esa enzima
        start = 0 #posición inicial
        while True: #bucle para buscar todo
            pos = secuencia_str.find(secuencia_recon, start)  # bucle para buscar todas las apariciones
            if pos == -1: # si no se encuentra más, salimos del bucle
                break
            posiciones.append(pos)  # guardamos la posición donde corta
            start = pos + 1 # avanzamos para seguir buscando más adelante
        if posiciones:  # si encontramos posiciones de corte
            enzimas_cortan[enzima] = posiciones # guardamos la enzima y las posiciones
        else: 
            enzimas_no_cortan.append(enzima) # si no corta, la añadimos a la lista de “no cortan”

    cortes = sorted([p for posiciones in enzimas_cortan.values() for p in posiciones]) # ordena todas las posiciones donde las enzimas cortan
    cortes = [0] + cortes + [len(secuencia_str)] # añade el inicio y el final de la secuencia
    fragmentos = [secuencia_str[cortes[i]:cortes[i+1]] for i in range(len(cortes)-1)] # crea los fragmentos de ADN entre cada corte

#imprimimos las que cortan
    print("Enzimas que cortan:") #muestra titulo para las enzimas que cortan
    for e, p in enzimas_cortan.items(): #recorre las enzimas que cortan
        print(f"  - {e}: posiciones {p}") #imprime el nombre de la enzima y sus posiciones de corte

#imprimimos las que no cortan
    print("\nEnzimas que no cortan:") #muestra titulo enzimas que no cortan
    if enzimas_no_cortan:
        print("  " + ", ".join(enzimas_no_cortan)) #las muestra separada por comas
    else:
        print("  Todas las enzimas cortan la secuencia.") #sino dice que todas las enzimas cortan

#imprimimos los resultados y los devolvemos para usar despues    
    print("\nFragmentos resultantes:")
    for i, frag in enumerate(fragmentos, 1):
        print(f"  Fragmento {i}: tamaño {len(frag)} pb  | Secuencia: {frag}")

    
    return {
        "enzimas_cortan": enzimas_cortan, #nos devuelve las que cortan
        "enzimas_no_cortan": enzimas_no_cortan, #nos devuelve las que no cortan
        "fragmentos": fragmentos #fragmentos resultantes del corte
    }

enzR = { #diccionario de enzimas y sus secuencias de corte
    'HindIII': 'A*AGCTT',
    'BamHI': 'G*GATCC',
    'AluI': 'AG*CT',
    'Sau3AI': '*GATC',
    'EcoRI': 'G*AATTC'
}

# llamamos a la funcion 
resultado = digestion_restriccion(secuencia, enzR)
```

#### 7.- Realiza una función que mediante la introducción de una secuencia de ADN devuelva lo siguiente: (3pto)
1. Todos los ORFs posibles que puedan surgir de la secuencia.
2. Selecciona los 6 ORFs más largos e indica si corresponden a la strand positiva o negativa.
3. Devuelve como mejor predicción, el más largo de todos.
4. Calcula el porcentaje de aminoácidos aromáticos de esa proteína elegida y su longitud relativa a la secuencia original de ADN.

```python
def Orfs(secuencia):
    secuencia_orf = "".join(secuencia).upper().replace(" ","").replace("\n","")#hacemos que la secuencia sea seguida
#tenemos que crear una hebra complementaria que lea de 3'-5', por lo que hacemos un diccionario y cambiamos los nucelotidos por sus complementarias
    complemento = {'A':'T', 'T':'A', 'C':'G', 'G':'C'}
    reversa_complementaria = "".join(complemento[b] for b in reversed(secuencia_orf)) #toma cada base de la secuencia al reves, cambiala por su complementaria y une todas las letras en una sola cadena
    frames = [] #creamos una lista para guardar los 6 frames posibles
    for i in range(3): #genera los 3 valores de los tres posibles desplazamientos
        frames.append((f"+{i+1}",secuencia_orf[i:])) #creamos el marco positivo de lectura
        frames.append((f"-{i+1}",reversa_complementaria[i:])) #creamos el marco negativo, que es complementaria e inversa a la original
    stop_codones = ["TAA", "TAG", "TGA"] #estos son los codones de parada, hasta aqui es el ORF
    orfs = [] #lista vacia con los ORF
    for strand, seq in frames: #para cada elemento dentro de frames, guardame el primer valor en strad y el segundo en seq
        for i in range(0, len(seq)-2, 3): #recorre la secuencia de 3 en 3 bases
            codon = seq[i:i+3] #si encuentra un ATG, detecta un posible inicio de ORF
            if codon == "ATG": #este bucle dice que si encunetras un ATG, busca el primer codon de parada
                for j in range(i, len(seq)-2, 3):
                    codon2 = seq[j:j+3]
                    if codon2 in stop_codones: #cuando lo encunetra, corta la secuencia desde ATG hasta el codon de parada incluido
                        orf_seq = seq[i:j+3]
                        orfs.append((strand, orf_seq))#guarda este ORF junto al frame donde lo encontro
                        break
                
    orfs.sort(key=lambda x: len(x[1]), reverse=True) #creamos una lista que se ordena mediante funciones, seleccionan la segunda parte de la tupla (secuencia) y se ordena por longitud
    top6 = orfs[:6]
    
    mejor_orf = top6[0] #nos quedamos con los 6 mas largos
    
    amino_acidos_aromaticos = ["F", "Y", "W"] #estos son los tres aa aromaticos
    #tabla de traduccion de cada codon a aminoacidos
    codon_tabla = {
        "TTT":"F","TTC":"F","TTA":"L","TTG":"L","CTT":"L","CTC":"L","CTA":"L","CTG":"L",
        "ATT":"I","ATC":"I","ATA":"I","ATG":"M","GTT":"V","GTC":"V","GTA":"V","GTG":"V",
        "TCT":"S","TCC":"S","TCA":"S","TCG":"S","CCT":"P","CCC":"P","CCA":"P","CCG":"P",
        "ACT":"T","ACC":"T","ACA":"T","ACG":"T","GCT":"A","GCC":"A","GCA":"A","GCG":"A",
        "TAT":"Y","TAC":"Y","TAA":"*","TAG":"*","CAT":"H","CAC":"H","CAA":"Q","CAG":"Q",
        "AAT":"N","AAC":"N","AAA":"K","AAG":"K","GAT":"D","GAC":"D","GAA":"E","GAG":"E",
        "TGT":"C","TGC":"C","TGA":"*","TGG":"W","CGT":"R","CGC":"R","CGA":"R","CGG":"R",
        "AGT":"S","AGC":"S","AGA":"R","AGG":"R","GGT":"G","GGC":"G","GGA":"G","GGG":"G"
    }
    
    proteina = "" #lista vacia para guardar la proteina
    seq_orf = mejor_orf[1] #cogemos la mejor secuencia
    for i in range(0, len(seq_orf)-2, 3): #recorre la secuencia de tres en tres (codones)
        codon = seq_orf[i:i+3] #extrae un triplete (codon)
        proteina += codon_tabla.get(codon, "X") #busca el codon en la tabla de aa y añade el correspondiente, y si no existe da X como error
    
    total_aa = len(proteina) #numero total de aa
    aromáticos = sum(proteina.count(a) for a in amino_acidos_aromaticos)#cuenta cuantas veces aparece uno aromatico
    porcentaje_aromaticos = (aromáticos / total_aa) * 100 if total_aa > 0 else 0 #saca el porcentaje de aromaticos
    longitud_relativa = len(seq_orf) / len(secuencia_orf) * 100 #dice que porcentaje del original ocupa este ORF
    #devuelve los resultados como un diccionario
    return {
    "Total_ORFs_encontrados": len(orfs),
    "Top6_ORFs_longitudes_nt": [len(seq) for _, seq in top6],
    "Mejor_ORF_strand": mejor_orf[0],
    "Mejor_ORF_longitud_nt": len(mejor_orf[1]),
    "Mejor_ORF_secuencia": mejor_orf[1],  
    "Proteina_traducida": proteina,
    "Proteina_traducida_longitud_aa": total_aa,
    "Porcentaje_aminoacidos_aromaticos": round(porcentaje_aromaticos, 2),
    "Longitud_relativa_porcentaje": round(longitud_relativa, 2)
}
resultado = Orfs(secuencia)
for clave, valor in resultado.items():
    print(f"{clave}: {valor}\n")
```
